# HVAC FaultSense: Notebook 01 - Dataset Profiling and Quality Audit

This notebook loads the official LBNL Rooftop Unit (RTU) dataset, generates the Brick Schema data dictionary, computes descriptive statistics, and visualizes physical sensor distributions.

```python
import pandas as pd
import numpy as np
from pathlib import Path
from hvac_faultsense.data.ingestion import load_simulated_dataset
from hvac_faultsense.data.metadata import build_dataset_manifest, get_data_dictionary
from hvac_faultsense.data.validation import profile_dataset, generate_profile_markdown
from hvac_faultsense.visualization.plots import plot_missing_values, plot_correlation_heatmap
```

## 1. Load Data Dictionary and Dataset Manifest

```python
base_dir = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
manifest = build_dataset_manifest(base_dir / 'data' / 'raw' / 'simulated')
data_dict = get_data_dictionary()
print(f'Cataloged {len(manifest)} scenarios.')
data_dict.head(10)
```

## 2. Ingest and Statistically Profile Telemetry

```python
df = load_simulated_dataset(base_dir / 'data' / 'raw' / 'simulated', stride=30)
print('Loaded shape:', df.shape)
profile = profile_dataset(df)
pd.DataFrame(profile['statistics']).T.head(15)
```

## 3. Correlation Heatmap

```python
corr_df = pd.DataFrame(profile['correlations'])
fig_corr = plot_correlation_heatmap(corr_df)
fig_corr.show()
```
